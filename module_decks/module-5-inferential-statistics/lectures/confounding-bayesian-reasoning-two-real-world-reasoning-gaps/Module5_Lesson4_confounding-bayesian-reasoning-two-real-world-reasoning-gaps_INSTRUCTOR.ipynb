### 🤔 Check for Understanding: Firefighters and Fire Damage

*Mode: Quick-Fire Game*

A study finds towns with more firefighters also have more fire damage. What's the real confounder?

- Firefighters cause fires to spread further
- Fire size drives both firefighter count and damage
- More fire damage attracts new firefighters to town

**Answer:** Fire size drives both firefighter count and damage — A bigger fire genuinely calls for more firefighters AND causes more damage - fire size drives both numbers at once. Neither firefighter count nor damage is causing the other; a bigger fire is the real confounder behind the correlation.

## Run the Real Update

*Context: `k` is how many observed penguins that year were Gentoo; `n` is the total observed that year - both real counts, read directly off the same dataset from the last two classes.*

**Instructor notes:**

> [ACTIVITY] Run it and read all three years' output before the next slide interprets it.
> 
> [SCRIPT] Name `.ppf` before moving on. It's the inverse of a cumulative distribution - plugging in .025 and .975 finds the two values bounding the middle 95% of the current belief. That range is what the next slide names formally.
> 
> [SCRIPT] Confirm the real numbers as they print: 2007 mean≈0.312, 2008 mean≈0.358, 2009 mean≈0.361 - and the 95% range narrowing each year.

*Input:*


```python
from scipy import stats

alpha, beta = 1, 1
batches = [(2007, 34, 110), (2008, 46, 114), (2009, 44, 120)]
for year, k, n in batches:
    alpha += k
    beta += n - k
    mean = alpha / (alpha + beta)
    ci = stats.beta.ppf([.025, .975], alpha, beta)
    print(year, round(mean, 3), ci.round(3))
```
